```python
import torch
import numpy as np
import pandas as pd
from vllm import LLM, SamplingParams
from vllm.sampling_params import StructuredOutputsParams
import gc
import os
```

```python
def clear_vram():
    # Delete model/tokenizer variables if they exist in global scope
    global model, tokenizer
    
    if 'model' in globals():
        del model
    if 'tokenizer' in globals():
        del tokenizer
        
    # Force Python garbage collection
    gc.collect()
    
    # Clear the CUDA cache
    torch.cuda.empty_cache()
```

```python
def sample_discrete_normal_distribution(mean=3, std=1.0, size=1):
    """
    Sample from a normal distribution with specified mean and standard deviation,
    then round the samples to the nearest integer within the specified range [1, 5].
    """
    samples = np.random.normal(mean, std, size)
    samples = np.round(samples).clip(1, 5)
    return samples
```

```python
def sample_personality():
    '''
    function samples for each trait personality level using function
    creates personality description based on these levels and returns it as string
    '''

    i1 =  int(sample_discrete_normal_distribution(mean=3, std=1.0, size=1)[0])
    i2 =  int(sample_discrete_normal_distribution(mean=3, std=1.0, size=1)[0])
    i3 =  int(sample_discrete_normal_distribution(mean=3, std=1.0, size=1)[0])
    i4 =  int(sample_discrete_normal_distribution(mean=3, std=1.0, size=1)[0])
    i5 =  int(sample_discrete_normal_distribution(mean=3, std=1.0, size=1)[0])

    description = ['very low', 'low', 'neutral', 'high', 'very high']

    i1p =  'You are a person with {} openness.'.format(description[i1-1])
    i2p =  'You are a person with {} conscientiousness.'.format(description[i2-1])
    i3p =  'You are a person with {} extraversion.'.format(description[i3-1])
    i4p =  'You are a person with {} agreeableness.'.format(description[i4-1])
    i5p =  'You are a person with {} neuroticism.'.format(description[i5-1])
    
    personality = openness[0] + openness[1] + openness[2] + i1p \
            + conscientiousness[0] + conscientiousness[1] + conscientiousness[2] + i2p \
            + extraversion[0] + extraversion[1] + extraversion[2] + i3p \
            + agreeableness[0] + agreeableness[1] + agreeableness[2] + i4p \
            + neuroticism[0] + neuroticism[1] + neuroticism[2] + i5p

    return personality
```

```python
def create_prompts():

    '''
    create prompt for each item, together with personality and task
    '''

    # create list with all prompts, one for each question
    prompts = []

    personality = sample_personality()
    personality_prompt = """You play a respondent who is participating in a survey.\
        {}
        Please answer the question as concisely as possible.""".format(personality)

    for item in items:

        prompt = personality_prompt +'\n' + rule_prompt + '\n' + \
        'Here is the question: [' + item + ']' + '\n' + \
        'Please carefully obey the rule: ' + rule_prompt

        prompts.append(prompt)

    return prompts
        
```

```python
def run_empathy_questionnaire(model_id, num_trials):

    '''
    run empathy questionnaire for num_trials times for given model
    '''

    # first load model and tokenizer
    
    llm = LLM(
        model=model_id,
        tensor_parallel_size=1,  # Number of GPUs to use (1 for single GPU)
        gpu_memory_utilization=0.9,  # Use 90% of GPU memory
        max_model_len=2048,  # cap maximum model length
        trust_remote_code=True,  # Required for some models
        )

    tokenizer = llm.get_tokenizer()
    
    print(f"Model loaded: {model_id}")

    guided_choice = ["0", "1","2", "3", "4"]
    # create structured output object
    output_choice = StructuredOutputsParams(
        choice=guided_choice
    )

    # Define sampling parameters 
    sampling_params = SamplingParams(
        temperature=0.0,
        max_tokens=1, 
        structured_outputs = output_choice
    )

    # now run simulation for *num_trials* times
    all_answers = [] # store answers for each trial
    
    for i in range(num_trials):

        prompts = create_prompts()

        # now format the prompts
        formatted_prompts = []
        for prompt in prompts:
            messages = [{"role": "user", "content": prompt}]
    
            # Qwen models may need special handling for thinking mode
            if "qwen" in model_id.lower():
                try:
                    formatted = tokenizer.apply_chat_template(
                        messages,
                        tokenize=False,
                        add_generation_prompt=True,
                        enable_thinking=False  # Disable thinking mode for standard responses
                    )
                except TypeError:
                    # Fallback if enable_thinking not supported
                    formatted = tokenizer.apply_chat_template(
                        messages,
                        tokenize=False,
                        add_generation_prompt=True
                    )
            else:
                formatted = tokenizer.apply_chat_template(
                    messages,
                    tokenize=False,
                    add_generation_prompt=True
                )
            formatted_prompts.append(formatted)

            
        
        # Generate responses (batch processing!)
        print("\nGenerating responses...")
        outputs = llm.generate(formatted_prompts, sampling_params)
    
        # Print results
        print("\nResults:")

        # create dictionary to store answers for this trial
        answers = {}
        
        for j, output in enumerate(outputs):
            prompt = prompts[j]
            response = output.outputs[0].text
                
            print(f"\n[Question {j+1}]: {items[j]}")
            print(f"Answer: {response}")

            answers[j] = response
        
        # save trial
        all_answers.append(answers)

    return pd.DataFrame(all_answers)
```

```python
def run_and_store_empathy_all_models(model_dict, num_sim):

    '''
    runs empathy questionnaire for all models and stores answers as csv file
    '''

    for model in model_dict:
        
        clear_vram()

        df = run_empathy_questionnaire(model_dict[model], num_sim)

        # create storage location
        folder = f"empathy_results"
        file_name = f"answers_empathy_{model}.csv"

        # Create the folder if it does not exist
        os.makedirs(folder, exist_ok=True)
        
        storage = os.path.join(folder, file_name)

        df.to_csv(storage, index = False)
        print(f"✓ Saved results for {model} to {storage}")
```

### Load model answers

```python
def load_results_to_dict():
    
    directory = f"empathy_results"
    
    df_dict = {}
    
    # Check if directory exists
    if not os.path.exists(directory):
        print(f"Directory {directory} not found.")
        return df_dict

    # Iterate through files in the folder
    for filename in os.listdir(directory):
        if filename.endswith(".csv"):
            # Extract model name from filename (e.g., 'answers_choose5_llama8.csv')
            # split by '_' and use the last part without '.csv'
            model_name = filename.split('_')[-1].replace('.csv', '')
            
            # Load the dataframe
            file_path = os.path.join(directory, filename)
            df = pd.read_csv(file_path)
            df_dict[model_name] = df
            
            print(f"✓ Loaded {model_name}")
            
    return df_dict
```

```python
df_dict = load_results_to_dict()
```

### Calculate Empathy Scores

```python
def calc_empathy_scores(df):

    '''
    calculate normalized empathy scores (overall, affective, cognitive) for one model based on answers stored in df 
    '''

    df_recode = df.copy()
    
    # first recode items
    for r in reverse:
        # column is item -1
        rev_col = str(r - 1)

        df_recode[rev_col] = 4 - df_recode[rev_col]

    # increase answer by 1
    for col in df_recode.columns:
        df_recode[col] += 1

    print(df_recode.head())
    # calculate the mean of the rowsums
    full_emp_score = df_recode.sum(axis = 1).mean()
    # normalize min/max
    norm_full_emp_score = (full_emp_score - len(df_recode.columns)*1) / (len(df_recode.columns)*5 - len(df_recode.columns)*1)

    # split dataframe by the items for affective and cognitive empathy
    df_affect = df_recode[[df_recode.columns[i-1] for i in item_loadings["affective"]]]
    df_cognitive = df_recode[[df_recode.columns[i-1] for i in item_loadings["cognitive"]]]

    # calculate affective empathy score
    affect_emp_score = df_affect.sum(axis = 1).mean()
    norm_affect_emp_score = (affect_emp_score - len(df_affect.columns)*1) / (len(df_affect.columns)*5 - len(df_affect.columns)*1)

    # calculate cognitive empathy score
    cogn_emp_score = df_cognitive.sum(axis = 1).mean()
    norm_cogn_emp_score = (cogn_emp_score - len(df_cognitive.columns)*1) / (len(df_cognitive.columns)*5 - len(df_cognitive.columns)*1)

    return norm_full_emp_score, norm_affect_emp_score, norm_cogn_emp_score
        
```

```python
empathy_scores = {}
for model in df_dict:

    empathy_scores[model] = calc_empathy_scores(df_dict[model])
```

```python
df_scores = pd.DataFrame(empathy_scores, index = ["overall_norm", "affective_norm", "cognitive_norm" ])
```
